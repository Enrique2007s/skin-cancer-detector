```python
KaggleDatasetPath = "rm1000/skin-cancer-isic-images"
DestinationFolder = "inputs/skin_cancer_dataset"   
! kaggle datasets download -d {KaggleDatasetPath} -p {DestinationFolder}
```

Now, we unzip the file and delete the zip file

```python
import zipfile
with zipfile.ZipFile(DestinationFolder + "/skin-cancer-isic-images.zip", 'r') as zip_ref:
    zip_ref.extractall(DestinationFolder)

os.remove(DestinationFolder + '/skin-cancer-isic-images.zip')
```
